```python
from zippermerge.differential import discrete
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
```

```python
t = [0, 0.1, 0.3, 0.4, 0.55, 0.67, 0.71]
x = [23.1, 22.5, 23.5, 21.88, 22.5, 23.5, 24.88]
```

```python
fig1 = plt.figure(figsize=(8, 5))
plt.plot(t, x, label="$x$", color="blue")

plt.xlabel("$t$")
plt.ylabel("$x$")
plt.title("x vs t")
plt.legend()
plt.grid(True)

plt.show()
```

```python
v = discrete.diff(t, x)
t_new = t[1:]

fig2 = plt.figure(figsize=(8, 5))
plt.plot(t_new, v, label="$v$", color="red")

plt.xlabel("$t$")
plt.ylabel("$v$")
plt.title("v vs t")
plt.legend()
plt.grid(True)

plt.show()
```

```python
with PdfPages("plots.pdf") as pdf:
    pdf.savefig(fig1)
    pdf.savefig(fig2)
```
